```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import LabelEncoder

# Step 1: 读取数据
data = pd.read_csv('/home/zhanyu/experiment/data-hh/my/train_86i3lia5Jj4=_aggregated_output.csv')

# Step 2: 编码分类变量
categorical_columns = ['flt_date', 'flt_no', 'a', 'b', 'c', 'aircraft']
label_encoders = {col: LabelEncoder() for col in categorical_columns}

for col in categorical_columns:
    data[col] = label_encoders[col].fit_transform(data[col])

# Step 3: 特征与目标分离
features = data[['flt_date', 'flt_no', 'a', 'b', 'c', 'aircraft', 'ab_duration', 'bc_duration', 'ac_duration']]
target_ab_pax = data['ab_pax']
target_bc_pax = data['bc_pax']
target_ac_pax = data['ac_pax']

# Step 4: 划分训练集和测试集
X_train, X_test, ab_pax_train, ab_pax_test = train_test_split(features, target_ab_pax, test_size=0.2, random_state=42)
_, _, bc_pax_train, bc_pax_test = train_test_split(features, target_bc_pax, test_size=0.2, random_state=42)
_, _, ac_pax_train, ac_pax_test = train_test_split(features, target_ac_pax, test_size=0.2, random_state=42)

# Step 5: 训练决策树模型
model_ab_pax = DecisionTreeRegressor(random_state=42)
model_ab_pax.fit(X_train, ab_pax_train)

model_bc_pax = DecisionTreeRegressor(random_state=42)
model_bc_pax.fit(X_train, bc_pax_train)

model_ac_pax = DecisionTreeRegressor(random_state=42)
model_ac_pax.fit(X_train, ac_pax_train)

# Step 6: 模型预测
ab_pax_pred = model_ab_pax.predict(X_test)
bc_pax_pred = model_bc_pax.predict(X_test)
ac_pax_pred = model_ac_pax.predict(X_test)

# Step 7: 模型评估
print("ab_pax Model Performance:")
print("MSE:", mean_squared_error(ab_pax_test, ab_pax_pred))
print("R²:", r2_score(ab_pax_test, ab_pax_pred))

print("\nbc_pax Model Performance:")
print("MSE:", mean_squared_error(bc_pax_test, bc_pax_pred))
print("R²:", r2_score(bc_pax_test, bc_pax_pred))

print("\nac_pax Model Performance:")
print("MSE:", mean_squared_error(ac_pax_test, ac_pax_pred))
print("R²:", r2_score(ac_pax_test, ac_pax_pred))

# Step 8: 展示预测结果与真实值对比
results_ab_pax = pd.DataFrame({
    'Actual_ab_pax': ab_pax_test,
    'Predicted_ab_pax': ab_pax_pred
}).reset_index(drop=True)

results_bc_pax = pd.DataFrame({
    'Actual_bc_pax': bc_pax_test,
    'Predicted_bc_pax': bc_pax_pred
}).reset_index(drop=True)

results_ac_pax = pd.DataFrame({
    'Actual_ac_pax': ac_pax_test,
    'Predicted_ac_pax': ac_pax_pred
}).reset_index(drop=True)

print("\nSample ab_pax Predictions:")
print(results_ab_pax.head())

print("\nSample bc_pax Predictions:")
print(results_bc_pax.head())

print("\nSample ac_pax Predictions:")
print(results_ac_pax.head())
```
