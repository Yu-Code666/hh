```python
import warnings
warnings.filterwarnings('ignore')
import argparse
import numpy as np
import networkx as nx
from gensim.models import Word2Vec
import random

import matplotlib.pyplot as plt
%matplotlib inline
```

## 读入命令行参数

```python
def parse_args():
    '''
    Parses the node2vec arguments.
    '''
    # 使用 argparse 模块定义并解析命令行参数
    # 创建一个 ArgumentParser 对象，用于描述程序及其参数
    parser = argparse.ArgumentParser(description="Run node2vec.")
    
    # 定义输入文件参数（默认值为 'karate.edgelist'）
    # 图的输入文件，通常是邻接表或边列表格式
    parser.add_argument('--input', nargs='?', default='karate.edgelist',
                        help='Input graph path')  # 输入图的路径
    
    # 定义输出文件参数（默认值为 'karate.emb'）
    # 用于存储生成的节点嵌入向量
    parser.add_argument('--output', nargs='?', default='karate.emb',
                        help='Embeddings path')  # 节点嵌入向量文件路径
    
    # 定义嵌入向量的维度（默认值为 128）
    parser.add_argument('--dimensions', type=int, default=128,
                        help='Number of dimensions. Default is 128.')  # 嵌入向量的维度大小
    
    # 定义随机游走的序列长度（默认值为 80）
    parser.add_argument('--walk-length', type=int, default=80,
                        help='Length of walk per source. Default is 80.')  # 每次随机游走的步长
    
    # 定义每个节点生成随机游走序列的次数（默认值为 10）
    parser.add_argument('--num-walks', type=int, default=10,
                        help='Number of walks per source. Default is 10.')  # 每个节点随机游走的次数
    
    # 定义 Word2Vec 的窗口大小（默认值为 10）
    # 用于优化上下文窗口
    parser.add_argument('--window-size', type=int, default=10,
                        help='Context size for optimization. Default is 10.')  # 上下文窗口大小
    
    # 定义 SGD（随机梯度下降）训练的迭代次数（默认值为 1）
    parser.add_argument('--iter', default=1, type=int,
                        help='Number of epochs in SGD')  # 随机梯度下降的迭代次数
    
    # 定义并行线程的数量（默认值为 8）
    # 用于多线程训练 Word2Vec
    parser.add_argument('--workers', type=int, default=8,
                        help='Number of parallel workers. Default is 8.')  # 并行线程数
    
    # 定义 Node2Vec 的 p 参数（默认值为 1）
    # 控制返回路径的偏向性，p 越高越倾向于回到上一节点
    parser.add_argument('--p', type=float, default=1,
                        help='Return hyperparameter. Default is 1.')  # 返回超参数 p
    
    # 定义 Node2Vec 的 q 参数（默认值为 2）
    # 控制前进路径的偏向性，q 越高越倾向于探索新节点
    parser.add_argument('--q', type=float, default=2,
                        help='Inout hyperparameter. Default is 2.')  # 前进超参数 q
    
    # 定义是否为加权图的参数
    # 如果设置 `--weighted`，图将被视为加权图
    parser.add_argument('--weighted', dest='weighted', action='store_true',
                        help='Boolean specifying (un)weighted. Default is unweighted.')  # 是否加权
    parser.add_argument('--unweighted', dest='unweighted', action='store_false')
    parser.set_defaults(weighted=False)  # 默认是无权图
    
    # 定义图是否为有向图的参数
    # 如果设置 `--directed`，图将被视为有向图
    parser.add_argument('--directed', dest='directed', action='store_true',
                        help='Graph is (un)directed. Default is undirected.')  # 是否为有向图
    parser.add_argument('--undirected', dest='undirected', action='store_false')
    parser.set_defaults(directed=False)  # 默认是无向图
    
    # 返回解析后的命令行参数
    return parser.parse_args(args=[])

# 调用 parse_args 函数解析参数
args = parse_args()
```

```python
def alias_setup(probs):
    '''
    Compute utility lists for non-uniform sampling from discrete distributions.
    Refer to https://hips.seas.harvard.edu/blog/2013/03/03/the-alias-method-efficient-sampling-with-many-discrete-outcomes/
    for details
    '''
    K = len(probs)
    # q corrsespond to Prob
    q = np.zeros(K)
    # J Alias
    J = np.zeros(K, dtype=np.int)

    smaller = []
    larger = []
    
    # 将各个概率分成两组，一组的概率值大于1，另一组的概率值小于1
    for kk, prob in enumerate(probs):
        q[kk] = K*prob # 每类事件的概率 乘 事件个数
        
        # 判定”劫富”和“济贫“的对象
        if q[kk] < 1.0:
            smaller.append(kk)
        else:
            larger.append(kk)
    
    # 使用贪心算法，将概率值小于1的不断填满
    # pseudo code step 3
    while len(smaller) > 0 and len(larger) > 0:
        small = smaller.pop()
        large = larger.pop()

        J[small] = large
        # 更新概率值，劫富济贫，削峰填谷
        q[large] = q[large] - (1 - q[small])
        if q[large] < 1.0:
            smaller.append(large) # 把被打倒的土豪归为贫农
        else:
            larger.append(large)

    return J, q
```

```python
def alias_draw(J, q):
    '''
    Draw sample from a non-uniform discrete distribution using alias sampling.
    O(1)的采样
    '''
    K = len(J) # 事件个数

    kk = int(np.floor(np.random.rand()*K)) # 生成1到K的随机整数
    if np.random.rand() < q[kk]:
        return kk # 取自己本来就对应的事件 
    else:
        return J[kk] # 取alias事件
```

```python
def get_alias_edge(src, dst):
    p = args.p
    q = args.q

    unnormalized_probs = []
    
    # 论文3.2.2节核心算法，计算各条边的转移权重
    for dst_nbr in sorted(G.neighbors(dst)):
        if dst_nbr == src:
            unnormalized_probs.append(G[dst][dst_nbr]['weight']/p)
        elif G.has_edge(dst_nbr, src):
            unnormalized_probs.append(G[dst][dst_nbr]['weight'])
        else:
            unnormalized_probs.append(G[dst][dst_nbr]['weight']/q)

    # 归一化各条边的转移权重
    norm_const = sum(unnormalized_probs)
    normalized_probs =  [float(u_prob)/norm_const for u_prob in unnormalized_probs]
    
    # 执行 Alias Sampling
    return alias_setup(normalized_probs)
```

```python
is_directed = args.directed
alias_nodes = {}

# 节点概率alias sampling和归一化
for node in G.nodes():
    unnormalized_probs = [G[node][nbr]['weight'] for nbr in sorted(G.neighbors(node))]
    norm_const = sum(unnormalized_probs)
    normalized_probs =  [float(u_prob)/norm_const for u_prob in unnormalized_probs]
    alias_nodes[node] = alias_setup(normalized_probs)
    # 信息展示
    if node == 25:
        print('25号节点')
        print(unnormalized_probs)
        print(norm_const)
        print(normalized_probs)
        print(alias_nodes[node])
```

## 生成一条随机游走序列

```python
def node2vec_walk(walk_length, start_node):
    '''
    从指定的起始节点，生成一个随机游走序列
    '''
    # 初始化游走序列，将起始节点加入序列
    walk = [start_node]

    # 循环，直到生成长度为 walk_length 的节点序列
    while len(walk) < walk_length:
        # 获取当前节点 cur 的邻居节点列表，并进行排序
        cur = walk[-1]
        cur_nbrs = sorted(G.neighbors(cur))

        # 如果当前节点有邻居，继续生成随机游走；否则结束游走
        if len(cur_nbrs) > 0:
            # 如果游走序列只有一个节点，从当前节点的邻居中使用别名采样选择下一个节点
            if len(walk) == 1:
                walk.append(cur_nbrs[alias_draw(alias_nodes[cur][0], alias_nodes[cur][1])])
            # 如果游走序列已有多个节点，从当前节点 cur 的邻居中使用别名采样选择下一个节点
            # 并考虑前一个节点 prev 的影响（Node2Vec 的策略）
            else:
                prev = walk[-2]
                next = cur_nbrs[alias_draw(alias_edges[(prev, cur)][0],
                    alias_edges[(prev, cur)][1])]
                walk.append(next)
        # 如果当前节点没有邻居节点，终止随机游走
        else:
            break

    # 返回生成的随机游走序列
    return walk
```

## 采样得到所有随机游走序列

```python
def simulate_walks(num_walks, walk_length):
    '''
    图中每个节点作为起始节点，生成 num_walk 个随机游走序列
    '''
    walks = []
    nodes = list(G.nodes())
    print('Walk iteration:')
    for walk_iter in range(num_walks):
        print(str(walk_iter+1), '/', str(num_walks))
        # 打乱节点顺序
        random.shuffle(nodes)
        for node in nodes:
            walks.append(node2vec_walk(walk_length=walk_length, start_node=node))

    return walks
```
