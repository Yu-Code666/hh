```python
import pandas as pd
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_squared_error
import joblib
```

## 加载数据

```python
# 读取Excel文件
df = pd.read_excel("/home/zhanyu/experiment/data-hh/航班销售过程数据202301-202406/2023-01-01/flight_sale_process_2023-01-01-1.xlsx")
```

```python
# 查看数据的前几行
print(df.head(60))
```

## 删除不需要的列

因为我们不关心公务舱的情况，只关心经济舱的情况，所以只需要经济舱的部分

```python
# 删除不需要的列
df = df.drop(columns=['tktRev', 'pax', 'cPax', 'cCap', 'cRev'])

# 查看结果
print(df.head())
```

## 处理日期字段

```python
# 确保 flightDate 字段是 datetime 类型
df['flightDate'] = pd.to_datetime(df['flightDate'])

# 提取年、月、日、星期等字段
df['year'] = df['flightDate'].dt.year
df['month'] = df['flightDate'].dt.month
df['day'] = df['flightDate'].dt.day
df['weekday'] = df['flightDate'].dt.weekday  # 返回星期几，0=星期一, 6=星期日
df['weekday_name'] = df['flightDate'].dt.strftime('%A')  # 返回星期几的名称，如 'Monday'

df = df.drop(columns=['flightDate'])

# 查看结果
print(df.head())
```

```python
# 确保 depTime 字段是 datetime 类型
df['depTime'] = pd.to_datetime(df['depTime'])

# 提取时和分
df['hour'] = df['depTime'].dt.hour
df['minute'] = df['depTime'].dt.minute

df = df.drop(columns=['depTime'])
# 查看结果
print(df.head())
```
