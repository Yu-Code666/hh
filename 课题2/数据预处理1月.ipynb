```python
import os
import pandas as pd
```

## 加载数据

```python
# 定义数据目录和条件
data_dir = "/home/zhanyu/experiment/data-hh/海航系航班销售过程数据202301-202406-新"
# flight_condition = 'A4tgQv5cqkc='
flight_condition = 'A4tgQv5cqkc='
A_condition = 'xIqe+QQIQXQ='
B_condition = 't+kFcRdzL6o='
C_condition = '/M8gHzjxpNU='
```

```python
# 初始化一个空的 DataFrame 用于存储结果
combined_df = pd.DataFrame()
```

```python
# 获取所有1月份的CSV文件
csv_files = [file for file in os.listdir(data_dir) if file.endswith('.csv') and file.startswith('2023-01')]

# 按文件名中的日期部分进行排序
csv_files.sort(key=lambda x: x.split('.')[0])  # 从文件名中提取日期并按字典顺序排序

# 遍历排序后的文件
for file_name in csv_files:
    file_path = os.path.join(data_dir, file_name)
    
    # 读取当前文件
    df = pd.read_csv(file_path)
    
    # 筛选满足条件的行
    filtered_df = df[
        (df['航班号'] == flight_condition) &
        (df['A'] == A_condition) &
        (df['B'] == B_condition) &
        (df['C'] == C_condition)
    ]
     # 对筛选后的数据按 '航班日期', 'A', 'B', 'C', '航段', '航班号', '距离天数' 分组，并保留人数最大的行
    filtered_df = filtered_df.loc[filtered_df.groupby(['航班日期', 'A', 'B', 'C', '航段', '航班号', '距离天数'])['人数'].idxmax()]
    
    # 将筛选结果追加到总的 DataFrame 中
    combined_df = pd.concat([combined_df, filtered_df], ignore_index=True)
    
    # 输出当前文件中满足条件的行数
    print(f"{file_name} 中读取到了 {len(filtered_df)} 行数据")

# 查看总的结果
print(f"总共读取到了 {len(combined_df)} 行数据")
```

```python
# 对 combined_df 数据框按 '航班日期' 升序排序，然后按 '距离天数' 降序排序，并直接修改原始数据框
combined_df.sort_values(by=['航班日期', '距离天数'], ascending=[True, False], inplace=True)

# 假设 df 是你的 DataFrame
combined_df.to_csv('2023-01-A4tgQv5cqkc=.csv', index=False, encoding='utf-8-sig')
```
