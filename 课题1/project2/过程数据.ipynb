```python
# 重新引入 glob 模块
import glob
import pandas as pd
import os
from tqdm import tqdm

import warnings
warnings.filterwarnings("ignore", category=UserWarning)


# 匹配所有xlsx文件路径
xlsx_files = glob.glob("../../../data-hh/海航系销售过程数据-2025/*.xlsx")

# 更换输出目录，使用当前目录下的 converted_csv 文件夹
output_dir = "./converted_csv"
os.makedirs(output_dir, exist_ok=True)

# 转换并显示进度条
for file in tqdm(xlsx_files, desc="正在转换为CSV", unit="文件"):
    df = pd.read_excel(file)
    filename = os.path.splitext(os.path.basename(file))[0] + ".csv"
    output_path = os.path.join(output_dir, filename)
    df.to_csv(output_path, index=False, encoding="utf-8-sig")

len(xlsx_files)
```

```python
import pandas as pd
import glob
import os
from tqdm import tqdm

# 目标CSV文件路径
output_file = "merged_sales_data.csv"

# 查找所有CSV文件
csv_files = glob.glob("./converted_csv/*.csv")

# 如果输出文件已存在则删除，确保是新建的
if os.path.exists(output_file):
    os.remove(output_file)

# 边读取边写入
first_file = True
for file in tqdm(csv_files, desc="边合并边写入CSV", unit="文件"):
    for chunk in pd.read_csv(file, chunksize=10000):
        chunk.to_csv(output_file, mode='a', index=False, header=first_file, encoding='utf-8-sig')
        first_file = False  # 之后都不写入header

output_file if not first_file else None
```
