```python
import os
import pandas as pd
from pathlib import Path

# 指定目录路径
xlsx_dir = '/home/zhanyu/data-hh/2025/haihangSalesProcess/海航系销售过程数据'

# 存储所有DataFrame的字典
dataframes = {}

# 遍历目录中的所有文件
for file in os.listdir(xlsx_dir):
    if file.endswith('.xlsx'):
        file_path = os.path.join(xlsx_dir, file)
        print(f"正在加载文件: {file}")
        try:
            # 读取xlsx文件到DataFrame
            df = pd.read_excel(file_path)
            # 将DataFrame存储在字典中，以文件名(不含扩展名)为键
            dataframes[os.path.splitext(file)[0]] = df
            print(f"成功加载文件 {file}，数据形状: {df.shape}")
        except Exception as e:
            print(f"加载文件 {file} 时出错: {str(e)}")

# 打印加载的文件数量
print(f"\n总共加载了 {len(dataframes)} 个xlsx文件")

# 简单展示一下第一个DataFrame的前几行（如果有的话）
if dataframes:
    first_key = list(dataframes.keys())[0]
    print(f"\n第一个文件 '{first_key}' 的前5行数据:")
    print(dataframes[first_key].head())
else:
    print("\n未找到任何xlsx文件")
```
