```python
# # 读取CSV文件
# # /Users/zhanyu/code/data-hh/海航系销售结果数据
# df = pd.read_csv('../../data-hh/海航系销售结果数据/hh_result_23-6-7.csv')

# 显示前两行数据以确保正确加载
# print(df.shape)
# print(df.head(5))

import pandas as pd
import glob
import os

# 指定目录路径
directory_path = '../../data-hh/海航系销售结果数据/'

# 使用 glob 匹配所有 CSV 文件
csv_files = glob.glob(os.path.join(directory_path, "*.csv"))

# 检查匹配到的 CSV 文件列表
print("找到的 CSV 文件:")
print(csv_files)

# 读取所有 CSV 文件并合并为一个 DataFrame
dataframes = []  # 用于存储每个 CSV 文件的 DataFrame
for file in csv_files:
    print(f"正在读取文件: {file}")
    df = pd.read_csv(file)
    dataframes.append(df)

# 合并所有 DataFrame（可根据需要选择拼接方式）
df = pd.concat(dataframes, ignore_index=True)

# 显示合并后的数据形状和前几行
print("合并后的数据形状:", df.shape)
print(df.head())
```

## 合并航线

```python
# 定义需要添加后缀的列
suffix_columns = ['dep_time', 'cap', 'duration', 'tkt_rev', 'pax','unit_price']
# 定义不需要改变的列
keep_columns = ['flt_date', 'flt_no', 'bd_type', 'aircraft', 'a', 'b', 'c']

# 定义leg_no到后缀的映射
leg_suffix_map = {
    1: 'ab',
    2: 'bc',
    3: 'ac'
}

# 创建存储结果和异常组的列表
result = []
abnormal_groups = []
skipped_groups = 0

# 按照flt_date和flt_no分组
for (flt_date, flt_no), group in df.groupby(['flt_date', 'flt_no']):
    if len(group) != 3:
        abnormal_info = {
            'flt_date': flt_date,
            'flt_no': flt_no,
            'row_count': len(group),
            'leg_numbers': group['leg_no'].tolist()
        }
        abnormal_groups.append(abnormal_info)
        skipped_groups += 1
        continue
        
    new_row = {}
    
    # 处理不变的列
    for col in keep_columns:
        new_row[col] = group[col].iloc[0]
    
    # 处理需要添加后缀的列
    for col in suffix_columns:
        sorted_group = group.sort_values('leg_no')
        for _, row in sorted_group.iterrows():
            leg_no = row['leg_no']
            suffix = leg_suffix_map[leg_no]  # 使用映射获取对应的后缀
            new_row[f'{col}_{suffix}'] = row[col]
    
    result.append(new_row)

# 转换结果为DataFrame
result_df = pd.DataFrame(result)
abnormal_df = pd.DataFrame(abnormal_groups)

# 打印统计信息
print(f"总分组数：{skipped_groups + len(result)}")
print(f"被跳过的分组数：{skipped_groups}")
print(f"保留的分组数：{len(result)}")

# 显示异常组的情况
print("\n异常组情况：")
print(abnormal_df)

# 保存结果
result_df.to_csv('../../data-hh/my/hh_result/hh_result_2024_6_merged.csv', index=False)
```

## 筛除pax_ab  pax_bc  pax_ac全为0的航班

```python
# 筛除全部客流为0的航班
filtered_df = result_df[
    ~((result_df['pax_ab'] == 0) & 
      (result_df['pax_bc'] == 0) & 
      (result_df['pax_ac'] == 0))
]

# 重置索引
filtered_df = filtered_df.reset_index(drop=True)

# 显示筛选结果
print(f"筛选后的航班数量: {len(filtered_df)}")
print(f"筛除的航班数量: {len(result_df) - len(filtered_df)}")
print(f"筛除比例: {((len(result_df) - len(filtered_df)) / len(result_df) * 100):.2f}%")

# 更新result_df
result_df = filtered_df
```
