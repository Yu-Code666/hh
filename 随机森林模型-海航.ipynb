```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_squared_error
import joblib

# 1. 加载数据
file_path = "/home/zhanyu/experiment/data-hh/my/train_86i3lia5Jj4=.csv"  # 替换为你的文件路径
data = pd.read_csv(file_path)

# 2. 数据预处理
# 从日期字段提取年份、月份和日期
data['flt_year'] = pd.to_datetime(data['flt_date']).dt.year
data['flt_month'] = pd.to_datetime(data['flt_date']).dt.month
data['flt_day'] = pd.to_datetime(data['flt_date']).dt.day

# 编码其他字符串字段
label_encoders = {}
for col in ['a', 'b', 'c', 'segment', 'flt_no', 'aircraft']:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col])
    label_encoders[col] = le

# 删除原始日期字段（如果不再需要）
data.drop(columns=['flt_date'], inplace=True)

# 处理缺失值（填充为均值）
data.fillna(data.mean(), inplace=True)

# 3. 特征和标签划分
X = data[['flt_year', 'flt_month', 'flt_day', 'a', 'b', 'c', 'segment', 'flt_no', 'aircraft', 'duration']]
y_y_pax = data['y_pax']
y_c_pax = data['c_pax']

# 4. 划分训练集和测试集
X_train, X_test, y_y_train, y_y_test = train_test_split(X, y_y_pax, test_size=0.2, random_state=42)
_, _, y_c_train, y_c_test = train_test_split(X, y_c_pax, test_size=0.2, random_state=42)

# 5. 训练随机森林模型
rf_y_pax = RandomForestRegressor(n_estimators=100, random_state=42)
rf_y_pax.fit(X_train, y_y_train)

rf_c_pax = RandomForestRegressor(n_estimators=100, random_state=42)
rf_c_pax.fit(X_train, y_c_train)

# 6. 模型评估
y_y_pred = rf_y_pax.predict(X_test)
y_c_pred = rf_c_pax.predict(X_test)

mse_y_pax = mean_squared_error(y_y_test, y_y_pred)
mse_c_pax = mean_squared_error(y_c_test, y_c_pred)

print(f"y_pax MSE: {mse_y_pax}")
print(f"c_pax MSE: {mse_c_pax}")

# 7. 保存模型
joblib.dump(rf_y_pax, "rf_y_pax_model.pkl")
joblib.dump(rf_c_pax, "rf_c_pax_model.pkl")

print("模型已保存！")

# 8. 展示部分预测结果
results = pd.DataFrame({
    'flt_year': X_test['flt_year'],
    'flt_month': X_test['flt_month'],
    'flt_day': X_test['flt_day'],
    'y_pax_actual': y_y_test,
    'y_pax_predicted': y_y_pred,
    'c_pax_actual': y_c_test,
    'c_pax_predicted': y_c_pred
})

# 打印前10条结果
print("部分预测结果：")
print(results.head(10))
```
