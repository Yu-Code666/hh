```python
import pandas as pd

# 假设这是原始数据
data = pd.read_csv("/home/zhanyu/experiment/data-hh/my/train_86i3lia5Jj4=.csv")

# 1. 计算航段的客流量 (pax) 和时长 (duration)
data["pax"] = data["y_pax"] + data["c_pax"]  # 总客流量

def calculate_values(segment, pax, duration):
    if segment == "zwVGiYamVc0=-FWD6O2fHsIU=":  # ab 段
        return "ab_pax", pax, "ab_duration", duration
    elif segment == "FWD6O2fHsIU=-wSF4qXvI044=":  # bc 段
        return "bc_pax", pax, "bc_duration", duration
    elif segment == "zwVGiYamVc0=-wSF4qXvI044=":  # ac 段
        return "ac_pax", pax, "ac_duration", duration
    else:
        return None, 0, None, 0

# 创建新列存储航段类型及其 pax 和 duration
data["segment_type_pax"], data["segment_pax"], data["segment_type_duration"], data["segment_duration"] = zip(
    *data.apply(lambda row: calculate_values(row["segment"], row["pax"], row["duration"]), axis=1)
)

# 2. 按 flt_date 和 flt_no 分组，聚合 pax 和 duration
aggregated_data = data.pivot_table(
    index=["flt_date", "flt_no", "a", "b", "c", "aircraft"],
    columns=["segment_type_pax", "segment_type_duration"],
    values=["segment_pax", "segment_duration"],
    aggfunc="sum"
).reset_index()

# 3. 调整列名，展开多级索引
aggregated_data.columns = [
    f"{col[1]}" if col[0] in ["segment_pax", "segment_duration"] else col for col in aggregated_data.columns
]

# 4. 重命名列
aggregated_data = aggregated_data.rename(
    columns={"ab_pax": "ab_pax", "bc_pax": "bc_pax", "ac_pax": "ac_pax", 
             "ab_duration": "ab_duration", "bc_duration": "bc_duration", "ac_duration": "ac_duration"}
)

# 5. 查看结果
print(aggregated_data.head())

# 保存结果（如果需要）
aggregated_data.to_csv("aggregated_output.csv", index=False)
```
